```python
import json
import math
from functools import reduce

# =============================================================================
# SCRIPT 1: IMPUESTO A LAS ACTIVIDADES ECONÓMICAS
# =============================================================================

def calcular_impuesto_parcial(monto):
    # Diccionario simulado mediante lista de tuplas con la tabla oficial del examen
    # Formato: (DESDE, HASTA, PRECIO, ADICIONAL)
    tabla_impuestos = [
        (0.01, 500.00, 1.5, 0),
        (500.01, 1000.00, 1.5, 3),
        (1000.01, 2000.00, 3, 3),
        (2000.01, 3000.00, 6, 3),
        (3000.01, 8000.00, 9, 2),
        (8000.01, 18000.00, 15, 2),
        (18000.01, 30000.00, 39, 2),
        (30000.01, 60000.00, 63, 1),
        (60000.01, 100000.00, 93, 0.8),
        (100000.01, 200000.00, 125, 0.7),
        (200000.01, 300000.00, 195, 0.6),
        (300000.01, 400000.00, 255, 0.45),
        (400000.01, 500000.00, 300, 0.4),
        (500000.01, 100000.00, 340, 0.30),
        (1000000.01, 99999999.00, 490, 0.18)
    ]

    if monto < 0.01:
        return 0.0

    # Lógica de cálculo según fórmula del PDF
    for desde, hasta, precio, adicional in tabla_impuestos:
        if desde <= monto <= hasta:
            if adicional == 0:
                return precio
            else:
                impuesto = ((monto - desde) / 1000) * adicional + precio
                return round(impuesto, 2)
    return 0.0

def menu_impuestos():
    while True:
        print("\n" + "="*55)
        print(" IMPUESTO A LAS ACTIVIDADES ECONÓMICAS (PARCIAL I)")
        print("="*55)

        try:
            entrada = input("Ingrese el monto de la actividad económica (o '0' para salir): $").strip()
            monto = float(entrada)

            if monto == 0:
                print("Cerrando sistema de impuestos...")
                break

            total = calcular_impuesto_parcial(monto)
            print(f">>> Valor a pagar: ${total:,.2f}")

        except ValueError:
            print("[!] Error: Ingrese un valor numérico válido.")


# =============================================================================
# SCRIPT 2: CONVERSOR DE ÁREA
# =============================================================================

def conversor_area_parcial(valor, clave_origen, clave_destino):
    # Diccionario de factores de conversión (Base: 1 Metro Cuadrado)
    factores = {
        "a": 0.092903,   # Pie Cuadrado
        "b": 0.698896,   # Vara Cuadrada (1 vara = 0.836m)
        "c": 0.836127,   # Yarda Cuadrada
        "d": 1.0,        # Metro Cuadrado
        "e": 436.81,     # Tareas (625 varas cuadradas)
        "f": 6988.96,    # Manzana (10,000 varas cuadradas)
        "g": 10000.0     # Hectárea
    }

    valor_en_m2 = valor * factores[clave_origen]
    resultado = valor_en_m2 / factores[clave_destino]
    return round(resultado, 4)

def menu_conversor():
    opciones_nombres = {
        "a": "Pie Cuadrado",
        "b": "Vara Cuadrada",
        "c": "Yarda Cuadrada",
        "d": "Metro Cuadrado",
        "e": "Tareas",
        "f": "Manzana",
        "g": "Hectárea"
    }

    while True:
        print("\n" + "="*45)
        print("   CONVERSOR DE ÁREA - EL SALVADOR")
        print("="*45)
        for letra, nombre in opciones_nombres.items():
            print(f"  {letra}. {nombre}")
        print("  h. Salir")
        print("="*45)

        origen = input("Seleccione la unidad de ORIGEN (a-g): ").strip().lower()
        if origen == 'h':
            break
        if origen not in opciones_nombres:
            print("[!] Opción inválida.")
            continue

        destino = input("Seleccione la unidad de DESTINO (a-g): ").strip().lower()
        if destino not in opciones_nombres:
            print("[!] Opción inválida.")
            continue

        try:
            cantidad = float(input("Ingrese la cantidad a convertir: "))
            resultado = conversor_area_parcial(cantidad, origen, destino)

            n_origen = opciones_nombres[origen]
            n_destino = opciones_nombres[destino]
            print(f"\n>>> RESULTADO: {cantidad} {n_origen}(s) = {resultado} {n_destino}(s)")
            input("\nPresione ENTER para continuar...")
        except ValueError:
            print("[!] Error: Ingrese una cantidad numérica.")


# =============================================================================
# SCRIPT 3: SISTEMA ACADÉMICO Y ESTADÍSTICO INTERACTIVO - UGB (CICLO IV)
# =============================================================================

class SistemaEstadistico:
    """Clase encargada del procesamiento estadístico utilizando programación funcional."""
    def __init__(self, datos, nombre_variable="Variable"):
        self.datos = datos
        self.nombre_variable = nombre_variable
        self.n = len(datos)

    def calcular_media(self):
        if self.n == 0: return 0
        # Uso de reduce y lambda como se vio en clase
        suma = reduce(lambda acum, num: acum + float(num), self.datos, 0)
        return suma / self.n

    def calcular_varianza(self):
        if self.n <= 1: return 0
        media_val = self.calcular_media()
        # Uso de reduce para la varianza
        suma_cuadrados = reduce(lambda acum, num: acum + (float(num) - media_val)**2, self.datos, 0)
        return suma_cuadrados / self.n

    def calcular_desviacion_estandar(self):
        return math.sqrt(self.calcular_varianza())

    def calcular_mediana(self):
        if self.n == 0: return 0
        datos_ordenados = sorted([float(x) for x in self.datos])
        mitad = self.n // 2
        if self.n % 2 == 0:
            return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2.0
        else:
            return datos_ordenados[mitad]

    def calcular_frecuencias(self):
        frecuencias = {}
        for dato in self.datos:
            frecuencias[dato] = frecuencias.get(dato, 0) + 1
        return frecuencias

    def calcular_moda(self):
        if self.n == 0: return []
        frecuencias = self.calcular_frecuencias()
        max_freq = max(frecuencias.values())
        return [clave for clave, valor in frecuencias.items() if valor == max_freq]

    def calcular_rango(self):
        if self.n == 0: return 0
        return max(self.datos) - min(self.datos)

    def mostrar_reporte(self):
        print(f"\n" + "=" * 55)
        print(f" ANÁLISIS ESTADÍSTICO: {self.nombre_variable.upper()}")
        print("=" * 55)
        print(f"  • Total Registros (N)         : {self.n}")
        print(f"  • Media Aritmética            : {round(self.calcular_media(), 2)}")
        print(f"  • Mediana                     : {round(self.calcular_mediana(), 2)}")
        print(f"  • Moda                        : {self.calcular_moda()}")
        print(f"  • Varianza                    : {round(self.calcular_varianza(), 2)}")
        print(f"  • Desviación Estándar/Típica  : {round(self.calcular_desviacion_estandar(), 2)}")
        print(f"  • Rango                       : {round(self.calcular_rango(), 2)}")
        print(f"  • Frecuencias Absolutas       : {self.calcular_frecuencias()}")
        print("=" * 55)


class SistemaAcademico:
    """Gestión de la base de datos de los estudiantes."""
    def __init__(self):
        self.estudiantes = []

    def cargar_datos_json(self, json_raw):
        self.estudiantes = json.loads(json_raw)

    def agregar_estudiante(self, nombre, materia, nota, edad, traslado, telefono):
        nuevo = {
            "nombre": nombre,
            "materia": materia,
            "nota": float(nota),
            "edad": int(edad),
            "traslado_min": int(traslado),
            "uso_telefono_hrs": float(telefono)
        }
        self.estudiantes.append(nuevo)
        print(f"\n¡Estudiante '{nombre}' registrado exitosamente en la base de datos!")

    def extraer_variable(self, clave):
        return [e[clave] for e in self.estudiantes if clave in e]

    def mostrar_tabla_alumnos(self):
        print(f"\n{'='*105}")
        print(f"{'REGISTRO DEL SISTEMA ACADÉMICO':^105}")
        print(f"{'='*105}")
        print(f"{'No.':<4} | {'Nombre':<20} | {'Materia':<25} | {'Nota':<5} | {'Edad':<5} | {'Traslado (min)':<15} | {'Teléfono (hrs)':<15}")
        print("-" * 105)
        for i, alumno in enumerate(self.estudiantes, 1):
            print(f"{i:<4} | {alumno['nombre']:<20} | {alumno['materia']:<25} | {alumno['nota']:<5} | {alumno['edad']:<5} | {alumno['traslado_min']:<15} | {alumno['uso_telefono_hrs']:<15}")
        print(f"{'='*105}\n")


def ejecutar_menu_principal():
    # Dataset con los datos REALES extraídos de WhatsApp
    json_dataset = '''[
        {"nombre": "Samuel Portillo", "materia": "Estadística Computacional", "nota": 10.0, "edad": 20, "traslado_min": 80, "uso_telefono_hrs": 6.0},
        {"nombre": "Carlos Turcios", "materia": "Admin. Bases de Datos II", "nota": 9.5, "edad": 20, "traslado_min": 25, "uso_telefono_hrs": 5.0},
        {"nombre": "Gilda González", "materia": "Circuitos Digitales", "nota": 9.3, "edad": 19, "traslado_min": 30, "uso_telefono_hrs": 5.9},
        {"nombre": "Saúl Jimenez", "materia": "Programación Comp. II", "nota": 10.0, "edad": 21, "traslado_min": 60, "uso_telefono_hrs": 5.0},
        {"nombre": "Edgar Ramirez", "materia": "Inglés Básico", "nota": 10.0, "edad": 21, "traslado_min": 120, "uso_telefono_hrs": 7.0},
        {"nombre": "Yeimy Parada", "materia": "Admin. Bases de Datos II", "nota": 9.5, "edad": 20, "traslado_min": 45, "uso_telefono_hrs": 3.0},
        {"nombre": "Miguel Gómez", "materia": "Programación Comp. III", "nota": 9.5, "edad": 19, "traslado_min": 15, "uso_telefono_hrs": 4.0},
        {"nombre": "Josué Funes", "materia": "Estadística Computacional", "nota": 9.5, "edad": 19, "traslado_min": 30, "uso_telefono_hrs": 5.0},
        {"nombre": "Jefferson Castillo", "materia": "Circuitos Digitales", "nota": 9.5, "edad": 19, "traslado_min": 15, "uso_telefono_hrs": 2.3},
        {"nombre": "Ramael Cordova", "materia": "Inglés Básico", "nota": 9.5, "edad": 22, "traslado_min": 7, "uso_telefono_hrs": 13.5},
        {"nombre": "Carlos Parada", "materia": "Programación Comp. III", "nota": 8.5, "edad": 25, "traslado_min": 10, "uso_telefono_hrs": 5.5},
        {"nombre": "Madison Valle", "materia": "Estadística Computacional", "nota": 10.0, "edad": 20, "traslado_min": 30, "uso_telefono_hrs": 9.0},
        {"nombre": "Jennifer Renderos", "materia": "Admin. Bases de Datos II", "nota": 9.6, "edad": 21, "traslado_min": 60, "uso_telefono_hrs": 3.0},
        {"nombre": "Diego Villafranco", "materia": "Circuitos Digitales", "nota": 9.0, "edad": 20, "traslado_min": 30, "uso_telefono_hrs": 8.0},
        {"nombre": "Flor Campos", "materia": "Estadística Computacional", "nota": 9.0, "edad": 20, "traslado_min": 6, "uso_telefono_hrs": 5.4},
        {"nombre": "Wilfredo Ortez", "materia": "Programación Comp. III", "nota": 9.5, "edad": 20, "traslado_min": 30, "uso_telefono_hrs": 5.0},
        {"nombre": "Kimberly Castro", "materia": "Programación Comp. III", "nota": 10.0, "edad": 18, "traslado_min": 20, "uso_telefono_hrs": 9.0}
    ]'''

    sistema = SistemaAcademico()
    sistema.cargar_datos_json(json_dataset)

    variables_map = {
        "1": ("nota", "Notas Académicas"),
        "2": ("edad", "Edades"),
        "3": ("traslado_min", "Tiempo de Traslado (Minutos)"),
        "4": ("uso_telefono_hrs", "Uso de Teléfono (Horas/Día)")
    }

    while True:
        print("\n" + "=" * 55)
        print("         MENÚ SISTEMA ACADÉMICO Y ESTADÍSTICO")
        print("=" * 55)
        print("  [1] Ver tabla de estudiantes")
        print("  [2] Calcular estadísticas de UNA variable específica")
        print("  [3] Calcular estadísticas de TODAS las variables")
        print("  [4] Registrar un nuevo estudiante (Input)")
        print("  [5] Salir del programa")
        print("=" * 55)

        opcion = input("Seleccione una opción (1-5): ").strip()

        if opcion == "1":
            sistema.mostrar_tabla_alumnos()

        elif opcion == "2":
            print("\n--- SELECCIONE LA VARIABLE A ANALIZAR ---")
            print("  1. Notas Académicas")
            print("  2. Edades")
            print("  3. Tiempo de Traslado a la UGB (Minutos)")
            print("  4. Uso de Teléfono (Horas/Día)")

            sub_opcion = input("Elija una variable (1-4): ").strip()
            if sub_opcion in variables_map:
                clave, nombre = variables_map[sub_opcion]
                datos = sistema.extraer_variable(clave)
                analizador = SistemaEstadistico(datos, nombre_variable=nombre)
                analizador.mostrar_reporte()
            else:
                print("\n[!] Opción de variable no válida.")

        elif opcion == "3":
            for clave, nombre in variables_map.values():
                datos = sistema.extraer_variable(clave)
                analizador = SistemaEstadistico(datos, nombre_variable=nombre)
                analizador.mostrar_reporte()

        elif opcion == "4":
            print("\n--- INGRESO DE NUEVO ESTUDIANTE ---")
            try:
                nombre = input("Nombre completo: ").strip()
                materia = input("Materia actual: ").strip()
                nota = float(input("Nota (0.0 - 10.0): "))
                edad = int(input("Edad: "))
                traslado = int(input("Tiempo de traslado en minutos: "))
                telefono = float(input("Horas de uso de teléfono al día: "))
                sistema.agregar_estudiante(nombre, materia, nota, edad, traslado, telefono)
            except ValueError:
                print("\n[!] Error: Asegúrese de ingresar valores numéricos válidos.")

        elif opcion == "5":
            print("\nCerrando el sistema estadístico... ¡Éxitos en la entrega!")
            break

        else:
            print("\n[!] Opción inválida. Intente de nuevo con un número del 1 al 5.")


# =============================================================================
# MENÚ GENERAL PRINCIPAL
# =============================================================================

def menu_general():
    while True:
        print("\n" + "=" * 55)
        print("               MENÚ PRINCIPAL GENERAL")
        print("=" * 55)
        print("  [1] Impuesto a las Actividades Económicas")
        print("  [2] Conversor de Área (El Salvador)")
        print("  [3] Sistema Académico y Estadístico (UGB)")
        print("  [4] Salir del Sistema General")
        print("=" * 55)

        opcion = input("Seleccione una opción (1-4): ").strip()

        if opcion == "1":
            menu_impuestos()
        elif opcion == "2":
            menu_conversor()
        elif opcion == "3":
            ejecutar_menu_principal()
        elif opcion == "4":
            print("\nSaliendo del programa principal... ¡Hasta luego!")
            break
        else:
            print("\n[!] Opción inválida. Intente de nuevo con un número del 1 al 4.")

if __name__ == "__main__":
    menu_general()
```
