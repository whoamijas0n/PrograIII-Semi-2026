```python
import datetime
fecha = datetime.date.today()

dia= fecha.day
mes = fecha.month
año = fecha.year

print(fecha, "dia ", dia, "mes ", mes, "año", año)



semana = ("lunes", "martes", "miercoles", "jueves", "viernes", "sabado", "domingo")
print("semana", semana, "Primer dia de la semana: ", semana[0], "ultimo dia de la semana: ", semana[-1])

diaSemana = semana[fecha.isoweekday()]
print("dia de la semana: ", diaSemana)
print("Hoy es ", diaSemana, dia,"del mes", mes, "del año", año)
```

Haciendo uso de las listas en python calcule la nota final de computo I



```python
notas = [
    #Computo 1
    ["lab I",8 , 0.3],
    ["lab II",9 , 0.3],
    ["lab III",10 , 0.4],

    #Computo 2
    ["lab I",8 , 0.3],
    ["lab II",9 , 0.3],
    ["lab III",10 , 0.4],

    #Computo 3
    ["lab I",8 , 0.3],
    ["lab II",9 , 0.3],
    ["lab III",10 , 0.4],
]
```

```python
def calcualrNF(nota, notas):
  return notas[1]*notas[2] + nota
```

```python
from functools import reduce
nf = reduce(calcualrNF, notas, 0) / 3
print("nota final: ", nf)
```
