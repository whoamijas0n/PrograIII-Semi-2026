```python
import json
import statistics
from collections import Counter

# 1. CLASES (Programación Orientada a Objetos)

class Estudiante:
    """Clase para representar a un estudiante y sus datos."""
    def __init__(self, nombre, materia, nota, edad, tiempo_traslado_ugb, horas_telefono):
        self.nombre = nombre
        self.materia = materia
        self.nota = nota
        self.edad = edad
        self.tiempo_traslado_ugb = tiempo_traslado_ugb # En minutos
        self.horas_telefono = horas_telefono

    def a_diccionario(self):
        """Convierte el objeto a un diccionario para facilitar su uso con JSON."""
        return self.__dict__


class SistemaEstadistico:
    """Clase encargada de realizar todos los cálculos estadísticos requeridos."""
    def __init__(self, datos_numericos):
        self.datos = datos_numericos

    def calcular_media(self):
        return round(statistics.mean(self.datos), 2)

    def calcular_mediana(self):
        return statistics.median(self.datos)

    def calcular_moda(self):
        # multimode devuelve una lista de las modas, previniendo errores si hay empates
        modas = statistics.multimode(self.datos)
        return modas if len(modas) > 1 else modas[0]

    def calcular_varianza(self):
        # Varianza muestral
        return round(statistics.variance(self.datos), 4)

    def calcular_desviacion_estandar(self):
        # Desviación estándar (muestral, n-1)
        return round(statistics.stdev(self.datos), 4)

    def calcular_desviacion_tipica(self):
        # Desviación típica (poblacional, n) para ilustrar la diferencia teórica
        return round(statistics.pstdev(self.datos), 4)

    def calcular_rango(self):
        return round(max(self.datos) - min(self.datos), 2)

    def calcular_frecuencia(self):
        # Devuelve un diccionario con el conteo de cada valor
        return dict(Counter(self.datos))

    def generar_reporte(self, nombre_variable):
        """Imprime un reporte completo en consola."""
        print(f"\n--- REPORTE ESTADÍSTICO PARA: {nombre_variable.upper()} ---")
        print(f"Media Aritmética:    {self.calcular_media()}")
        print(f"Mediana:             {self.calcular_mediana()}")
        print(f"Moda(s):             {self.calcular_moda()}")
        print(f"Varianza:            {self.calcular_varianza()}")
        print(f"Desviación Estándar: {self.calcular_desviacion_estandar()} (Muestral)")
        print(f"Desviación Típica:   {self.calcular_desviacion_tipica()} (Poblacional)")
        print(f"Rango:               {self.calcular_rango()}")
        print(f"Frecuencias:         {self.calcular_frecuencia()}")


class SistemaAcademico:
    """Sistema principal que gestiona el dataset y las operaciones."""
    def __init__(self):
        self.dataset = []

    def cargar_datos(self, lista_estudiantes):
        """Agrega una lista de objetos Estudiante al sistema."""
        for est in lista_estudiantes:
            self.dataset.append(est)

    def exportar_json(self):
        """Genera un string JSON a partir de los diccionarios de estudiantes."""
        lista_diccionarios = [est.a_diccionario() for est in self.dataset]
        return json.dumps(lista_diccionarios, indent=4, ensure_ascii=False)

    def extraer_columna(self, llave):
        """Extrae todos los valores numéricos de una columna específica para analizarlos."""
        return [est.a_diccionario()[llave] for est in self.dataset]

# 2. DATASET (20 datos reales/simulados)

datos_compañeros = [
    Estudiante("Jason", "PROGRA III", 8.5, 21, 20, 5.0),
    Estudiante("Jaime", "Bases de Datos", 9.2, 20, 35, 6.5),
    Estudiante("Carlos", "PROGRA III", 7.5, 22, 45, 4.0),
    Estudiante("Jimmy", "PROGRA III", 8.0, 21, 15, 7.0),
    Estudiante("Mileydi", "PROGRA III", 6.8, 23, 60, 3.5),
    Estudiante("Roberto", "PROGRA III", 9.5, 19, 10, 8.0),
    Estudiante("Saul", "Progra III", 8.5, 24, 50, 4.5),
    Estudiante("Sharon", "Progra III", 7.0, 20, 25, 5.0),
    Estudiante("Flor", "Progra III", 7.5, 22, 30, 6.0),
    Estudiante("Jefferson", "Bases de Datos", 8.8, 21, 40, 5.5),
    Estudiante("Ramael", "Progra III", 9.0, 23, 15, 3.0),
    Estudiante("Jose", "Progra III", 8.2, 20, 55, 7.5),
    Estudiante("Luis", "Progra III", 6.5, 25, 20, 4.0),
    Estudiante("Miguel", "Progra III", 9.1, 19, 35, 6.5),
    Estudiante("Angel", "Progra III", 8.5, 21, 45, 5.0), # Moda intencional en nota (8.5)
    Estudiante("Jennifer", "Progra III", 7.8, 22, 10, 8.5),
    Estudiante("Herbert", "Progra III", 8.5, 20, 25, 6.0),
    Estudiante("Gloria", "Progra III", 9.0, 21, 30, 5.5),
    Estudiante("Diego", "Progra III", 7.2, 24, 60, 4.0),
    Estudiante("Saith", "Progra III", 8.9, 20, 15, 7.0)
]

# 3. EJECUCIÓN E INTERFAZ CON EL USUARIO

if __name__ == "__main__":
    # Inicializar el sistema académico y cargar datos
    sistema_academico = SistemaAcademico()
    sistema_academico.cargar_datos(datos_compañeros)

    # Imprimir un resumen del JSON generado usando diccionarios
    print("=== DATASET ACADÉMICO (Formato JSON) ===")
    json_data = sistema_academico.exportar_json()
    # Imprimimos solo los primeros 500 caracteres para no saturar la pantalla
    print(json_data[:500] + "\n... [Dataset truncado para visualización] ...\n")

    # Extraer variables para el análisis estadístico
    notas = sistema_academico.extraer_columna("nota")
    tiempos_ugb = sistema_academico.extraer_columna("tiempo_traslado_ugb")

    # Instanciar el sistema estadístico y generar reportes
    estadisticas_notas = SistemaEstadistico(notas)
    estadisticas_notas.generar_reporte("Notas de los estudiantes")

    estadisticas_tiempo = SistemaEstadistico(tiempos_ugb)
    estadisticas_tiempo.generar_reporte("Tiempo de traslado a la UGB (minutos)")
```
